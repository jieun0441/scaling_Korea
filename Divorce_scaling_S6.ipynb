```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Data Preparation
years = np.array([2007, 2008, 2009, 2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024])
elderly_idx = np.array([100, 102, 105, 107, 110, 113, 116, 120, 123, 126, 130, 133, 138, 142, 146, 151, 156, 161])
physicians_idx = np.array([100, 105, 110, 114, 114, 115, 119, 120, 123, 126, 127, 128, 130, 130, 130, 134, 136, 136])

# 2. Plot Configuration
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)
ax.grid(True, linestyle='--', alpha=0.5, zorder=0)
ax.set_facecolor('#fdfdfd')

# 3. Drawing Lines
line1, = ax.plot(years, elderly_idx, color='#e63946', linewidth=2.5, marker='o', markersize=4,
                  label='Elderly Population (Medical Demand)', zorder=4)
line2, = ax.plot(years, physicians_idx, color='#1d3557', linewidth=2.5, marker='s', markersize=4,
                  label='Number of Physicians (Medical Supply)', zorder=4)

# 4. Highlight the Supply-Demand Gap
ax.fill_between(years, elderly_idx, physicians_idx, where=(elderly_idx >= physicians_idx),
                 color='#e63946', alpha=0.12, label='Supply-Demand Gap', zorder=3)

# 5. Labels and Formatting
ax.set_xlabel('Year', fontsize=12, fontweight='bold', labelpad=10)
ax.set_ylabel('Index (2007 = 100)', fontsize=12, fontweight='bold', labelpad=10)
ax.set_xticks(years)
ax.set_xticklabels(years, rotation=45)

# [수정] 데이터 라벨이 테두리에 잘리지 않도록 우측 여백을 2025.5로 확장
ax.set_xlim(2006.5, 2025.5)
ax.set_ylim(95, 170)

# [수정] 텍스트 위치를 여백 공간에 맞춰 자연스럽게 배치
ax.text(2024.3, elderly_idx[-1], f"{elderly_idx[-1]}%", color='#e63946', fontweight='bold', va='center', ha='left')
ax.text(2024.3, physicians_idx[-1], f"{physicians_idx[-1]}%", color='#1d3557', fontweight='bold', va='center', ha='left')

# Title and Legend
#plt.title('The Illusion of Medical Growth in Shrinking Cities\n'
#          'Healthcare Supply Fails to Match Rapidly Aging Population (2007-2024)',
#          fontsize=14, fontweight='bold', pad=20, loc='left', color='#222222')
ax.legend(loc='upper left', frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()

# [추가] 300 DPI 고화질 JPG 파일로 저장 (bbox_inches='tight'로 라벨 잘림 방지)
plt.savefig('medical_growth_illusion.jpg', dpi=300, format='jpg', bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = np.array([2007, 2008, 2009, 2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024])
ratio_elderly_per_md = np.array([89.2, 86.6, 85.1, 84.4, 85.4, 87.6, 87.0, 88.7, 89.0, 89.4, 91.1, 92.9, 94.4, 97.7, 100.2, 100.6, 102.4, 105.6])

# Plot Configuration
fig, ax = plt.subplots(figsize=(10, 5), dpi=150)
ax.grid(True, linestyle='--', alpha=0.5, zorder=0)
ax.set_facecolor('#fdfdfd')

# Line Plot
ax.plot(years, ratio_elderly_per_md, color='#1d3557', linewidth=2.5, marker='d', markersize=5, zorder=4)

# [수정] 2007년 기준선을 눈에 잘 띄는 확실한 색상의 '실선'으로 변경
ax.axhline(ratio_elderly_per_md[0], color='#d90429', linestyle=':', linewidth=1.5, alpha=0.9, zorder=2)

# [수정] '2007 Level' 문구를 선과 겹치지 않게 2007년 점의 '위쪽 공간'으로 이동
ax.text(2007.0, ratio_elderly_per_md[0] + 2.3, '2007 Level (89.2)', color='#d90429', fontsize=10, fontweight='bold', ha='left')

# Labels and Formatting
ax.set_xlabel('Year', fontsize=12, fontweight='bold', labelpad=10)
ax.set_ylabel('Number of Elderly per Physician', fontsize=12, fontweight='bold', labelpad=10)
ax.set_xticks(years)
ax.set_xticklabels(years, rotation=45)

# [수정] 105.6 라벨이 우측 테두리에 잘리지 않도록 X축 범위를 2025.5로 확장
ax.set_xlim(2006.5, 2025.5)
ax.set_ylim(80, 110)

# [수정] 끝 수치(105.6)가 안쪽 여백에 안정적으로 위치하도록 ha='left' 지정
ax.text(2024.3, ratio_elderly_per_md[-1], f"{ratio_elderly_per_md[-1]}",  color='#1d3557', fontweight='bold', va='center', ha='left')

# Title
#plt.title('Growing Burden on Regional Healthcare\n'
#          'Trend of Elderly Population (Aged 65+) per Single Physician in Shrinking Cities',
#          fontsize=14, fontweight='bold', pad=20, loc='left', color='#222222')

plt.tight_layout()

# [추가] 300 DPI 고화질 JPG 파일로 저장
plt.savefig('elderly_per_physician_burden.jpg', dpi=300, format='jpg', bbox_inches='tight')

plt.show()
```
